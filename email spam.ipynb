```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df= pd.read_csv('mail_data.csv')
print(df)
```

```python

data = df.where(pd.notnull(df), '')

print(data.head(10))
```

```python
data.info()
```

```python
data.shape
```

```python
data.loc[data['Category'] == 'spam', 'Category',] =0
data.loc[data['Category'] == 'ham', 'Category',] =1
```

```python
X = data['Message']
```

```python
print(X)
```

```python
Y = data['Category']
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state = 3)
```

```python
print(X.shape)
print(X_train.shape)
print(X_test.shape)
```

```python
print(Y.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

feature_extraction = TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)

X_train_features = feature_extraction.fit_transform(X_train)  # fit on training data
X_test_features = feature_extraction.transform(X_test)        # transform test data

Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```

```python
print(X_train)
```

```python
print(X_train_features)
```

```python

from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_features, Y_train)
```

```python
prediction_on_training_data = model.predict(X_train_features)
accuracy_on_training_data = accuracy_score(Y_train, prediction_on_training_data)
```

```python
print('Acc on training data :', accuracy_on_training_data)
```

```python
prediction_on_testing_data = model.predict(X_test_features)
accuracy_on_testing_data = accuracy_score(Y_test, prediction_on_testing_data)
```

```python
print('Acc on training data :', accuracy_on_testing_data)
```

```python
input_your_mail = ["This is the second tme we have tried to contact u. U have won the RS40000 prize.2 claims is easy, just call 087104711148 Now! only 10p per mintue. BT-national-rate"]
input_data_features = feature_extraction.transform(input_your_mail)
prediction = model.predict(input_data_features)

print(prediction)

if(prediction[0]==1):
    print('Ham mail')
else:
    print('Spam mail')
```

```python
input_your_mail = ["Well obivously nit because all people in my cool college life went home;"]
input_data_features = feature_extraction.transform(input_your_mail)
prediction = model.predict(input_data_features)

print(prediction)

if(prediction[0]==1):
    print('Ham mail')
else:
    print('Spam mail')
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print("Accuracy:", accuracy_score(Y_test, prediction_on_testing_data))
print("Precision:", precision_score(Y_test, prediction_on_testing_data))
print("Recall:", recall_score(Y_test, prediction_on_testing_data))
print("F1 Score:", f1_score(Y_test, prediction_on_testing_data))
```
